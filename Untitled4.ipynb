```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# =============================================================================
# LATIHAN 1: Eksplorasi Data & Statistika Deskriptif dengan Pandas
# =============================================================================
print("--- Latihan 1: Eksplorasi Data & Statistika Deskriptif ---")

# (Ganti 'penjualan_nusantaramart.csv' dengan path dataset yang sesuai)
try:
    df = pd.read_csv('penjualan_nusantaramart.csv')

    print(df.head()) # lihat 5 baris pertama
    print(df.info()) # tipe data & jumlah data kosong
    print(df.describe()) # ringkasan statistik

    # Rata-rata total pembelian per kategori produk
    rata_kategori = df.groupby('kategori_produk')['total'].mean()
    print("\nRata-rata Pembelian per Kategori:\n", rata_kategori.sort_values(ascending=False))

except FileNotFoundError:
    print("[!] File CSV tidak ditemukan. Pastikan file berada di direktori yang sama.")

"""
JAWABAN TUGAS ANALISIS 1:
1. Kategori dengan rata-rata transaksi tertinggi:
   -> Dapat dilihat dari output `rata_kategori.sort_values(ascending=False)`. Kategori yang berada di urutan pertama (paling atas) adalah jawabannya.
2. Mengecek data kosong (missing value):
   -> Dari output `df.info()`, perhatikan kolom "Non-Null Count". Jika angkanya lebih kecil dari total baris data (RangeIndex), berarti ada missing value di kolom tersebut.
3. Perbandingan Mean dan Median kolom 'total':
   -> Cek output `df.describe()`. Bandingkan nilai `mean` dengan `50%` (median).
   -> Jika Mean > Median: Sebaran data 'right-skewed' (menjulur ke kanan). Artinya, ada beberapa transaksi dengan nilai sangat besar yang menarik rata-rata ke atas (umum terjadi pada data penjualan).
   -> Jika Mean = Median: Sebaran data normal/simetris.
"""

# Simulasi pembuatan DataFrame dummy jika CSV tidak ada agar kode visualisasi tetap bisa jalan
if 'df' not in locals():
    import numpy as np
    np.random.seed(42)
    dates = pd.date_range(start='2023-01-01', periods=100, freq='D')
    df = pd.DataFrame({
        'tanggal': np.random.choice(dates, 100),
        'kategori_produk': np.random.choice(['Sembako', 'Elektronik', 'Pakaian', 'Snack'], 100),
        'jumlah': np.random.randint(1, 20, 100),
        'rating_pelanggan': np.random.randint(1, 6, 100)
    })
    # Asumsi harga satuan
    harga = {'Sembako': 50000, 'Elektronik': 500000, 'Pakaian': 150000, 'Snack': 15000}
    df['total'] = df.apply(lambda row: row['jumlah'] * harga[row['kategori_produk']], axis=1)


# =============================================================================
# LATIHAN 2: Visualisasi Perbandingan dengan Bar Chart
# =============================================================================
total_kategori = df.groupby('kategori_produk')['total'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x=total_kategori.values, y=total_kategori.index, palette='Blues_r')
plt.title('Total Penjualan per Kategori Produk')
plt.xlabel('Total Penjualan (Rp)')
plt.ylabel('Kategori Produk')
plt.tight_layout()
plt.show()

"""
JAWABAN TUGAS ANALISIS 2:
1. Kategori penyumbang tertinggi dan terendah:
   -> Tertinggi: Bar yang berada di posisi paling atas (paling panjang).
   -> Terendah: Bar yang berada di posisi paling bawah (paling pendek).
2. Rekomendasi bisnis:
   -> Fokuskan stok dan kampanye marketing (seperti bundle/diskon) pada kategori unggulan untuk memaksimalkan profit.
   -> Untuk kategori terendah, manajer perlu mengevaluasi: apakah harganya terlalu mahal, penempatannya kurang strategis, atau perlukah produk tersebut didiskontinu untuk menghemat biaya gudang.
"""


# =============================================================================
# LATIHAN 3: Visualisasi Tren dengan Line Chart
# =============================================================================
df['tanggal'] = pd.to_datetime(df['tanggal'])
df['bulan'] = df['tanggal'].dt.to_period('M')
tren_bulanan = df.groupby('bulan')['total'].sum()

plt.figure(figsize=(8,5))
plt.plot(tren_bulanan.index.astype(str), tren_bulanan.values, marker='o', color='crimson')
plt.title('Tren Penjualan Bulanan')
plt.xlabel('Bulan')
plt.ylabel('Total Penjualan (Rp)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

"""
JAWABAN TUGAS ANALISIS 3:
1. Bulan dengan penjualan tertinggi & penyebabnya:
   -> Dilihat dari puncak tertinggi (peak) pada line chart. Penyebab umumnya di ritel Indonesia adalah momen Hari Raya (Idul Fitri/Natal), akhir tahun, atau masa gajian dengan promo besar (misal 11.11 atau 12.12).
2. Tren penjualan (naik/turun/fluktuatif):
   -> Biasanya fluktuatif dari bulan ke bulan, namun perhatikan garis lurus imajiner dari awal ke akhir (trendline). Jika ujung kanan lebih tinggi dari ujung kiri secara konsisten, maka tren cenderung NAIK (positif).
"""


# =============================================================================
# LATIHAN 4: Visualisasi Hubungan dengan Scatter Plot
# =============================================================================
plt.figure(figsize=(7,5))
sns.scatterplot(data=df, x='jumlah', y='total', hue='kategori_produk')
plt.title('Hubungan Jumlah Pembelian dengan Total Nilai Transaksi')
plt.xlabel('Jumlah Barang')
plt.ylabel('Total (Rp)')
plt.tight_layout()
plt.show()

"""
JAWABAN TUGAS ANALISIS 4:
1. Pola hubungan:
   -> Terdapat hubungan POSITIF. Semakin banyak jumlah barang yang dibeli (sumbu X ke kanan), maka total nilai transaksi (sumbu Y ke atas) akan semakin besar.
2. Titik outlier:
   -> Outlier dapat dilihat pada titik yang "menyendiri" dari kelompok garis utama.
   -> Misalnya, jika ada titik dengan jumlah barang sedikit tapi totalnya sangat tinggi (kemungkinan kategori Elektronik/Barang Mewah). Atau jumlah barang sangat banyak tapi total transaksinya rendah (kemungkinan kategori Snack/Barang Murah).
"""


# =============================================================================
# LATIHAN 5: Visualisasi Distribusi & Deteksi Outlier (Histogram & Boxplot)
# =============================================================================
# Histogram distribusi rating pelanggan
plt.figure(figsize=(7,5))
sns.histplot(df['rating_pelanggan'], bins=5, color='seagreen')
plt.title('Distribusi Rating Pelanggan')
plt.xlabel('Rating')
plt.ylabel('Frekuensi')
plt.show()

# Boxplot jumlah pembelian per kategori (deteksi outlier)
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='kategori_produk', y='jumlah')
plt.title('Sebaran Jumlah Pembelian per Kategori Produk')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

"""
JAWABAN TUGAS ANALISIS 5:
1. Distribusi rating:
   -> Dilihat dari tinggi batang histogram. Jika batang menumpuk di angka 4 dan 5, berarti distribusinya cenderung TINGGI (left-skewed/negatively skewed). Jika relatif rata di semua angka, distribusinya merata.
2. Outlier pada boxplot:
   -> Titik-titik hitam (diamond/lingkaran) yang berada di luar "kumis" (whiskers) boxplot adalah outlier.
   -> Penyebabnya: Pembelian dalam skala besar yang tidak biasa (misalnya oleh tengkulak, agen grosir, atau reseller) atau pelanggan korporat.
"""


# =============================================================================
# LATIHAN 6: Menggabungkan Grafik dalam Dashboard Sederhana
# =============================================================================
fig, axes = plt.subplots(2, 2, figsize=(13,9))

sns.barplot(x=total_kategori.values, y=total_kategori.index, ax=axes[0,0], palette='Blues_r')
axes[0,0].set_title('Total Penjualan per Kategori')

axes[0,1].plot(tren_bulanan.index.astype(str), tren_bulanan.values, marker='o', color='crimson')
axes[0,1].set_title('Tren Penjualan Bulanan')
axes[0,1].tick_params(axis='x', rotation=45)

sns.histplot(df['rating_pelanggan'], bins=5, ax=axes[1,0], color='seagreen')
axes[1,0].set_title('Distribusi Rating Pelanggan')

sns.boxplot(data=df, x='kategori_produk', y='jumlah', ax=axes[1,1])
axes[1,1].set_title('Sebaran Jumlah per Kategori')
axes[1,1].tick_params(axis='x', rotation=20)

plt.tight_layout()
plt.show()

"""
JAWABAN TUGAS ANALISIS 6:
1. Keuntungan Dashboard:
   -> Memberikan gambaran (helicopter view) yang komprehensif. Manajer dapat menghubungkan satu insight dengan insight lainnya secara instan. Misalnya, mengaitkan turunnya tren bulanan (kuadran 2) dengan distribusi rating yang rendah (kuadran 3).
2. Keputusan Bisnis Manajer:
   -> Mengalokasikan budget promosi lebih banyak pada masa menjelang bulan-bulan tren puncak.
   -> Membuka program khusus "B2B / Reseller" bagi pelanggan-pelanggan outlier (membeli dalam jumlah sangat banyak di boxplot) agar mereka menjadi pelanggan loyal.
   -> Mengevaluasi layanan atau produk jika terdapat lonjakan pada rating rendah.
"""
```
